```python
df=df.drop_duplicates()
```

```python
sns.boxplot('Shortness_of_Breath')
```

```python
x=df.drop('Heart_Risk',axis=1)
```

```python
y=df['Heart_Risk']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.8,random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
SC=StandardScaler()
x_train=SC.fit_transform(x_train)
x_test=SC.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
LR=LogisticRegression()
LR.fit(x_train,y_train)
y_pred=LR.predict(x_test)
```

```python
from sklearn.metrics import classification_report,accuracy_score
```

```python
print(classification_report(y_test,y_pred))
```

```python
accuracy_score(y_test,y_pred)
```

```python
from sklearn.model_selection import StratifiedKFold,cross_val_score
```

```python
skf=StratifiedKFold(n_splits=5,shuffle=True,random_state=42)
```

```python
cv_scores = cross_val_score(
    LR,
    x_train,
    y_train,
    cv=skf,
    scoring='accuracy'
)

print("5-Fold Accuracy Scores:", cv_scores)
print("Mean Accuracy:", cv_scores.mean())
```

```python
param_grid = {
    'C': [0.01, 0.1, 1, 10, 100],
    'solver': ['liblinear', 'lbfgs']
}
```

```python
from sklearn.model_selection import GridSearchCV
```

```python
grid = GridSearchCV(LR,param_grid,cv=5,scoring='accuracy')

grid.fit(x_train, y_train)
y_pred_grid=grid.predict(x_test)
print("Best Parameters:", grid.best_params_)
print("Best CV Accuracy:", grid.best_score_)
```

```python
print(classification_report(y_test, y_pred_grid))
```

```python
from sklearn.model_selection import RandomizedSearchCV
```

```python
param_random = {
    'C': [0.001, 0.01, 0.1, 1, 10, 100, 1000],
    'solver': ['liblinear', 'lbfgs']
}
```

```python
random_search = RandomizedSearchCV(LR,param_random,n_iter=5,cv=5,scoring='accuracy',random_state=42)

random_search.fit(x_train, y_train)
```

```python
print("Best Parameters:", random_search.best_params_)
print("Best CV Accuracy:", random_search.best_score_)
```

```python
y_pred_random=random_search.predict(x_test)
```

```python
print(classification_report(y_test, y_pred_random))
```

```python
from sklearn.ensemble import RandomForestClassifier

RF = RandomForestClassifier(
    random_state=42
)

RF.fit(x_train, y_train)

y_pred_rf = RF.predict(x_test)

print(classification_report(y_test, y_pred_rf))
```

```python
rf_scores = cross_val_score(
    RF,
    x_train,
    y_train,
    cv=5,
    scoring='accuracy'
)

print("Random Forest CV Scores:", rf_scores)
print("Random Forest Mean Accuracy:", rf_scores.mean())
```

```python
from sklearn.metrics import accuracy_score, precision_score
from sklearn.metrics import recall_score, f1_score
import pandas as pd

# Baseline prediction
y_pred_baseline = LR.predict(x_test)

comparison = pd.DataFrame({
    'Model': [
        'Baseline Logistic Regression',
        'GridSearchCV Logistic Regression',
        'RandomizedSearchCV Logistic Regression',
        'Random Forest'
    ],

    'Accuracy': [
        accuracy_score(y_test, y_pred_baseline),
        accuracy_score(y_test, y_pred_grid),
        accuracy_score(y_test, y_pred_random),
        accuracy_score(y_test, y_pred_rf)
    ],

    'Precision': [
        precision_score(y_test, y_pred_baseline),
        precision_score(y_test, y_pred_grid),
        precision_score(y_test, y_pred_random),
        precision_score(y_test, y_pred_rf)
    ],

    'Recall': [
        recall_score(y_test, y_pred_baseline),
        recall_score(y_test, y_pred_grid),
        recall_score(y_test, y_pred_random),
        recall_score(y_test, y_pred_rf)
    ],

    'F1 Score': [
        f1_score(y_test, y_pred_baseline),
        f1_score(y_test, y_pred_grid),
        f1_score(y_test, y_pred_random),
        f1_score(y_test, y_pred_rf)
    ]
})

comparison
```
